# 12. What is the distribution of movie releases over the years?

```python
file['release_date'] = pd.to_datetime(file['release_date'], errors='coerce')

years_20 = pd.to_datetime('today').year - 20
file_temp = file[file['release_date'].dt.year >= years_20]

file_temp['years'] = file_temp['release_date'].dt.year
file_temp['month'] = file_temp['release_date'].dt.month #to extract the year and month from the release_date.

file_temp['year'] = file_temp['release_date'].dt.to_period('Y')

release_counts = file_temp['year'].value_counts().sort_index()
print(release_counts)

file_temp = file_temp.pivot_table(index='years', columns='month', aggfunc='size', fill_value=0)

plt.figure(figsize=(12, 8))  # Adjusting the figure size
sns.heatmap(file_temp, cmap='YlGnBu', annot=True, fmt='d', cbar=True, linewidth=4) # fmt='d' rounding off any decimal points.

plt.title("Distribution of Movie Releases Over the Last 20 Years", fontsize=16)
plt.xlabel("Month", fontsize=14)
plt.ylabel("Year", fontsize=14)

plt.show()
```

##### Observation : 
* The number of movies released each year has been increasing steadily from 2005 to 2019, with the highest number of releases in 2019 (14,494 movies). After 2019, the number of releases went down a bit in 2020 and 2021. In 2024 the number dropped even more, and there are very few releases expected in 2025, 2027, and 2029 possibly due to incomplete data or future projections.

# Feature Engineering :

* In the world of data analytics and machine learning, raw data alone is rarely enough to build effective predictive models. Often, the true power of a dataset is unlocked not by the quantity of data, but by the quality and relevance of its features. This is where Feature Engineering comes into play.

* Feature Engineering is the process of selecting, modifying, and creating new features from raw data to increase the predictive power of machine learning algorithms. It’s a crucial step in the data pipeline, acting as the bridge between data collection and model training. Done well, it can significantly boost the performance of a model — sometimes even more than choosing a more complex algorithm.

# Why is Feature Engineering Important?

##### Machine learning models learn patterns from data. But if the data is not structured in a way that highlights those patterns, the model will struggle to learn effectively. Feature engineering:

* Enhances signal and reduces noise.

* Injects domain knowledge into the dataset.

* Improves model accuracy and interpretability.

* Simplifies complex relationships by deriving meaningful variables.

##### In real-world datasets, especially those that are collected from business operations, sensor logs, customer transactions, or media platforms, raw features often hide valuable insights. Feature engineering helps reveal those insights by reshaping the data in more useful ways.

## 4. Gradient Boosting Regressor :
**Concept :**
* A boosting technique that improves predictions by sequentially correcting errors.

**How it works :**

* Start with a weak model
* Calculate errors
* Train a new model on those errors
* Repeat the process
* Final prediction = sum of all small models

**Best For :**
* Achieving high accuracy in structured datasets.

**Limitations:**

* Slow training time.
* Sensitive to hyperparameter tuning.
